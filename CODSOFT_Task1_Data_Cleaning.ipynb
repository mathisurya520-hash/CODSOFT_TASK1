```python

import pandas as pd
df = pd.read_csv('/content/retail_sales_dataset.csv')
print("Dataset imported successfully!")
print("\nFirst 5 rows:")
print(df.head())
```

```python
print("Dataset Shape:")
print(df.shape)
print("\nColumn Names:")
print(df.columns)
print("\nDataset Information:")
df.info()
```

```python
print("Missing Values in Each Column:")
print(df.isnull().sum())
print("\nTotal Missing Values:")
print(df.isnull().sum().sum())
```

```python
print("Total Duplicate Rows:")
print(df.duplicated().sum())
print("\nDuplicate Transaction IDs:")
print(df["Transaction ID"].duplicated().sum())
```

```python
# Convert Date column to datetime
df["Date"] = pd.to_datetime(df["Date"], errors="coerce")
print("Updated Data Types:")
print(df.dtypes)
print("\nDate Column Type:")
print(df["Date"].dtype)
print("\nInvalid Dates:")
print(df["Date"].isnull().sum())
```

```python
print("Gender Values:")
print(df["Gender"].value_counts())
print("\nProduct Categories:")
print(df["Product Category"].value_counts())
print("\nAge Range:")
print(df["Age"].min(), "to", df["Age"].max())
print("\nQuantity Range:")
print(df["Quantity"].min(), "to", df["Quantity"].max())
print("\nNegative or Zero Values:")
print("Age <= 0:", (df["Age"] <= 0).sum())
print("Quantity <= 0:", (df["Quantity"] <= 0).sum())
print("Price <= 0:", (df["Price per Unit"] <= 0).sum())
print("Total Amount <= 0:", (df["Total Amount"] <= 0).sum())
```

```python
# Final validation
print("Missing values:", df.isnull().sum().sum())
print("Duplicate rows:", df.duplicated().sum())
print("Duplicate Transaction IDs:",
      df["Transaction ID"].duplicated().sum())
# Check calculated amounts
incorrect_amounts = (
    df["Total Amount"] != df["Quantity"] * df["Price per Unit"]
).sum()
print("Incorrect total amounts:", incorrect_amounts)
# Save cleaned dataset
df.to_csv("retail_sales_cleaned.csv", index=False)
print("\nCleaned dataset saved successfully!")
print("Final shape:", df.shape)
```
